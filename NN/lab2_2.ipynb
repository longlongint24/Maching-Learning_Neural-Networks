```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

#generowanie danych systetycznych
x= 2* np.random.rand(100,1) #100 losowych punktow dla x z zakresu [0,2]
noise = np.random.randn(100,1)*0.2 #niewielki szum o rozdkladzie normalnym 
y= 2*x+1+noise
```

```python
w= np.random.randn(1)
b=np.random.randn(1)

learning_rate =  0.1 #szybkosc uczenia
epochs = 1000 #liczba iteracji petli
n = len(x) #liczba probek danych

loss_history=[]
```

```python
#pętla treningowa

for epoch in range(epochs):
    y_hat = w*x+b# obliczanie predykcji

    error = y_hat - y #obliczanie mse r2
    mse_loss= (1/n) * np.sum(error**2) 
    loss_history.append(mse_loss) 

    #pochodna mse wzgeldem 'w'
    dw= (2/n) * np.sum(x*error)
    #pochodna mse wzgledem 'b'
    db = (2/n) * np.sum(error)

    #update parametrow
    w = w-learning_rate *dw
    b= b-learning_rate *db

    if(epoch+1) %100 ==0:
        print(f"Epoka {epoch+1}: Loss = {mse_loss:.4f}, w = {w[0]:.4f}, b = {b[0]:.4f}")

print("\n--- Zakończono trening ---")
print(f"Wyuczone parametry: w = {w[0]:.4f}, b = {b[0]:.4f}")
print("Oczekiwane parametry blisko: w = 2.0, b = 1.0")
```

```python
plt.figure(figsize=(10, 5))
plt.plot(range(epochs), loss_history, color='blue', label='MSE Loss')
plt.title('Spadek funkcji kosztu (MSE) w trakcie treningu')
plt.xlabel('Epoka')
plt.ylabel('Wartość MSE')
plt.legend()
plt.grid(True)
plt.show()
```
