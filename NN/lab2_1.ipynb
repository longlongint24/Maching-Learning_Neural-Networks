```python
import torch
import numpy as np
```

```python
zeros = torch.zeros((2,3))
ones = torch.ones((2,3))
rand = torch.rand((2,3))

print(zeros)
print(ones)
print(rand)
```

```python
a = torch.tensor([[1., 2.], [3., 4.]])
b = torch.tensor([[5.,6.], [7., 8.]])

print(a)
print(b)
print(a+b)
print(a*b) #mnozenie elementowe
print(a @ b.T) #iloczyn macierzowy
```

```python
t= torch.arange(12).reshape(3,4)
print(t)
print(t.reshape(2,6)) #nowy ksztalt, kopijuje dane
print(t.view(2,6)) #ten sam tensor ale inny widok, nie kopijuje
```

```python
arr = np.array([1,2,3])
t1 = torch.tensor(arr) #kopia, nie widac updatu bo stara kopia
t2 = torch.as_tensor(arr) # wspoldzielona pamiec, czyli widac update
arr[0]=67
print(t1,t2)
```

```python
def tensor_operations_demo(t):
    mean_val = t.mean()
    std_val = t.std()

    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    t_device = t.to(device)

    print(f"Średnia: {mean_val:.4f}, Odchylenie: {std_val:.4f}")
    print(f"Tensor znajduje się na urządzeniu: {t_device.device}")

    return mean_val,std_val,t_device

my_tensor = torch.randn(100) #tworzy tensor 100 losowych liczb
tensor_operations_demo(my_tensor)
```
